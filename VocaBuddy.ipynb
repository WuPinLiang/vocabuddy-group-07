```python
import random

# 單字庫結構：鍵為英文單字，值包含 'translation' (中文解釋) 與 'category' (分類)
vocabulary = {
    "apple": {"translation": "蘋果", "category": "水果"},
    "book": {"translation": "書本", "category": "物品"},
    "run": {"translation": "跑步", "category": "動作"},
    "beautiful": {"translation": "美麗的", "category": "形容詞"},
    "python": {"translation": "程式語言/蟒蛇", "category": "名詞"}
}

# 初始化答題紀錄
score_record = {
    "correct": 0,
    "incorrect": 0,
    "mistakes": []  # 記錄曾經答錯的單字，方便日後複習
}

print("="*40)
print("單字庫與測驗系統初始化完成！")
print(f"目前系統內共有 {len(vocabulary)} 個單字。")
print("="*40)
```

```python
def show_info():
    """顯示單字總數與分類"""
    print("\n" + "=" * 40)
    print(f"目前單字總數：{len(vocabulary)} 個")
    categories = set(info["category"] for info in vocabulary.values())
    print(f"包含的單字分類有：{', '.join(categories)}")
    print("=" * 40 + "\n")


def random_word():
    """隨機抽取並顯示一個單字"""
    word = random.choice(list(vocabulary.keys()))
    info = vocabulary[word]
    print("\n" + "-" * 40)
    print("【隨機單字學習】")
    print(f"英文單字： {word}")
    print(f"中文解釋： {info['translation']}")
    print(f"單字分類： {info['category']}")
    print("-" * 40 + "\n")


def quiz():
    """簡單的單字測驗 (中翻英)"""
    word = random.choice(list(vocabulary.keys()))
    info = vocabulary[word]
    print("\n" + "*" * 40)
    print("【隨機單字測驗】")
    print(f"請問「{info['translation']}」的英文單字是什麼？")

    answer = input("請輸入你的答案 (輸入 q 可退出測驗): ").strip().lower()

    if answer == "q":
        print("取消測驗。")
        return

    if answer == word:
        print("答對了！太棒了！")
        score_record["correct"] += 1
    else:
        print(f"答錯了... 正確答案是：【 {word} 】")
        score_record["incorrect"] += 1
        if word not in score_record["mistakes"]:
            score_record["mistakes"].append(word)
    print("*" * 40 + "\n")


def show_records():
    """顯示答對與答錯紀錄"""
    print("\n" + "=" * 40)
    print("【學習成效紀錄】")
    print(f"累積答對次數： {score_record['correct']} 次")
    print(f"累積答錯次數： {score_record['incorrect']} 次")

    # 計算答對率
    total_attempts = score_record["correct"] + score_record["incorrect"]
    if total_attempts > 0:
        win_rate = (score_record["correct"] / total_attempts) * 100
        print(f"答對率： {win_rate:.1f}%")

    if score_record["mistakes"]:
        print(f"\n建議多加複習的單字：")
        for mistake in score_record["mistakes"]:
            print(f"   - {mistake} ({vocabulary[mistake]['translation']})")
    print("=" * 40 + "\n")


def main_menu():
    """主程式迴圈"""
    while True:
        print("========== 歡迎來到單字學習小幫手 ==========")
        print(" [1] 顯示單字總數與分類")
        print(" [2] 隨機抽取單字學習")
        print(" [3] 進行單字測驗 (中翻英)")
        print(" [4] 查看答題紀錄與錯題本")
        print(" [5] 離開程式")
        print("============================================")

        choice = input("請輸入數字選擇功能 (1-5): ").strip()

        if choice == "1":
            show_info()
        elif choice == "2":
            random_word()
        elif choice == "3":
            quiz()
        elif choice == "4":
            show_records()
        elif choice == "5":
            print("\n程式結束！祝你英文突飛猛進，下次見！")
            break
        else:
            print("\n無效的輸入，請確認您輸入的是 1 到 5 之間的數字。\n")


# 啟動主程式
main_menu()
```
